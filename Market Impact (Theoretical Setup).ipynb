```python
import numpy as np
import matplotlib.pyplot as plt
from src.execution_engine import AlmgrenChrissOptimizer

# Parameters
X_total = 100000
N_steps = 20
S0 = 100
opt = AlmgrenChrissOptimizer(X_total, N_steps, 0.2, 2.5e-7, 2.5e-6, 1/N_steps)

# Compare TWAP (Linear) vs. "Fast" Selling
n_twap = np.full(N_steps, X_total/N_steps)
n_fast = np.zeros(N_steps); n_fast[0:5] = X_total/5

_, p_twap = opt.simulate_path(n_twap, S0)
_, p_fast = opt.simulate_path(n_fast, S0)

plt.figure(figsize=(10,5))
plt.plot(p_twap, label='TWAP Execution (Low Impact)')
plt.plot(p_fast, label='Aggressive Execution (High Impact)')
plt.title("Impact of Trading Speed on Realized Price")
plt.ylabel("Price ($)")
plt.legend()
plt.show()
```
